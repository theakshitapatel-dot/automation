```python
import pandas as pd
import matplotlib.pyplot as plt

plt.style.use('ggplot')
```

```python
df = pd.read_csv("/content/Data_set 2 - Copy.csv")

print("Dataset Loaded Successfully")
```

TASK 5: REASONS FOR INVESTMENT

```python
reason = df['Reason_Equity'].value_counts()

print(reason)

reason.plot(kind='pie', autopct='%1.1f%%')

plt.ylabel("")

plt.show()
```

TASK 6: SAVING OBJECTIVES

```python
saving = df['What are your savings objectives?'].value_counts()

print(saving)

saving.plot(kind='bar')

plt.show()
```

 TASK 7: COMMON INFORMATION SOURCES

```python
source = df['Source'].value_counts()

print(source)

source.plot(kind='bar')

plt.show()
```
